```python
import pandas as pd
df = pd.read_csv('/content/sample_data/california_housing_train.csv')

print("First 5 records:")
display(df.head())

print("\nDataset information:")
df.info()

print("\nMissing values in each column:")
display(df.isnull().sum().to_frame("Missing_Count"))

print("\nNumber of completely duplicated rows:")
print(df.duplicated().sum())

print("\nOriginal dataset dimensions:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df_clean = df.copy()

duplicates_before = df_clean.duplicated().sum()

df_clean = df_clean.drop_duplicates()

duplicates_after = df_clean.duplicated().sum()

print("Duplicate rows before cleaning:", duplicates_before)
print("Duplicate rows after cleaning:", duplicates_after)
print("Shape after duplicate removal:", df_clean.shape)

display(df_clean)
```

```python
numeric_columns = df_clean.select_dtypes(include=['number']).columns.tolist()
categorical_columns = df_clean.select_dtypes(exclude=['number']).columns.tolist()

# Fill missing numerical values with the median
for column in numeric_columns:
    if column in df_clean.columns:
        df_clean[column] = df_clean[column].fillna(df_clean[column].median())

# Fill missing categorical values with the mode
for column in categorical_columns:
    if column in df_clean.columns:
        mode_values = df_clean[column].mode(dropna=True)
        if not mode_values.empty:
            df_clean[column] = df_clean[column].fillna(mode_values.iloc[0])
        else:
            df_clean[column] = df_clean[column].fillna("Unknown")

print("Missing values handled successfully!")
print("\nMissing values after imputation:")
display(df_clean.isnull().sum().to_frame("Missing_Count"))
```

```python
text_columns = df_clean.select_dtypes(
    include=["object", "string"]
).columns

for column in text_columns:
    df_clean[column] = (
        df_clean[column]
        .astype("string")
        .str.strip()
        .str.lower()
    )

print("Text standardization completed!")

display(df_clean)
```

```python
print("========== DATA QUALITY REPORT ==========")

print("\nOriginal shape:", df.shape)
print("Cleaned shape:", df_clean.shape)

print("\nRemaining duplicate rows:")
print(df_clean.duplicated().sum())

print("\nTotal missing values:")
print(df_clean.isnull().sum().sum())

print("\nMissing values by column:")
display(df_clean.isnull().sum().to_frame("Missing_Count"))

print("\nData types after cleaning:")
display(df_clean.dtypes.to_frame("Data_Type"))

print("\nValidation completed!")
```

```python
print("Numerical summary statistics:")
numerical_summary = df_clean.describe()
display(numerical_summary)

# Save numerical report
numerical_summary.to_csv("numerical_summary.csv")

# Check if there are any object (categorical) columns before describing them
categorical_cols = df_clean.select_dtypes(include=["object"]).columns
if len(categorical_cols) > 0:
    print("\nCategorical summary statistics:")
    categorical_summary = df_clean.describe(include="object")
    display(categorical_summary)
    categorical_summary.to_csv("categorical_summary.csv")
else:
    print("\nNo categorical columns found. Skipping categorical summary.")

print("\nSummary statistics saved successfully!")
```

```python
# Cell 9: Export cleaned dataset

df_clean.to_csv("cleaned_sales.csv", index=False)

print("Cleaned dataset saved as cleaned_sales.csv")
print("Raw dataset saved as raw_sales.csv")

print("\nCleaned dataset preview:")
display(df_clean.head())

print("\nFinal dataset shape:", df_clean.shape)
```
